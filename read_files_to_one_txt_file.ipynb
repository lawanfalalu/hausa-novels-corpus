```python
import os
from pathlib import Path
import PyPDF2

directories = [
    "txt_files",
    "pdf_files",
    "document_files",
]

def read_text_file(file_path):
    """Read content from a text file."""
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            return f.read()
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

def read_pdf_file(file_path):
    """Read content from a PDF file."""
    try:
        with open(file_path, 'rb') as f:
            pdf_reader = PyPDF2.PdfReader(f)
            text = ""
            for page in pdf_reader.pages:
                text += page.extract_text()
            return text
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

def read_directories(dirs):
    """Read all text files from specified directories."""
    all_contents = {}
    
    for directory in dirs:
        if not os.path.exists(directory):
            print(f"Directory '{directory}' not found.")
            continue
        
        all_contents[directory] = {}
        
        for filename in os.listdir(directory):
            file_path = os.path.join(directory, filename)
            
            if filename.endswith('.txt'):
                content = read_text_file(file_path)
            elif filename.endswith('.pdf'):
                content = read_pdf_file(file_path)
            else:
                continue
            
            if content:
                all_contents[directory][filename] = content
    
    return all_contents

if __name__ == "__main__":
    contents = read_directories(directories)
    
    for directory, files in contents.items():
        print(f"\n{directory}:")
        for filename, content in files.items():
            print(f"  - {filename}: {len(content)} characters")
            
            
```

```python
# Append all file contents into a single txt file
output_file = "all_files.txt"

with open(output_file, 'w', encoding='utf-8') as outfile:
    for directory, files in contents.items():
        outfile.write(f"\n{'='*50}\n")
        outfile.write(f"Directory: {directory}\n")
        outfile.write(f"{'='*50}\n\n")
        
        for filename, content in files.items():
            outfile.write(f"\n--- File: {filename} ---\n")
            outfile.write(content)
            outfile.write("\n")

print(f"All files have been appended to '{output_file}'")
```

```python
import re

# Read the combined file and clean it
with open(output_file, 'r', encoding='utf-8') as infile:
    raw_text = infile.read()

# Remove common artifacts
cleaned_text = raw_text

# Remove page number indicators (e.g., "Page 1", "page 2")
cleaned_text = re.sub(r'page\s+\d+', '', cleaned_text, flags=re.IGNORECASE)
cleaned_text = re.sub(r'Page\s+\d+', '', cleaned_text)

# Remove extra blank lines (replace multiple newlines with single newline)
cleaned_text = re.sub(r'\n\s*\n+', '\n', cleaned_text)

# Remove BOM characters (zero-width no-break space)
cleaned_text = cleaned_text.replace('\ufeff', '')

# Remove extra whitespace at the beginning and end
cleaned_text = cleaned_text.strip()

# Write cleaned content to a new file
cleaned_output_file = "all_files_cleaned.txt"
with open(cleaned_output_file, 'w', encoding='utf-8') as outfile:
    outfile.write(cleaned_text)

print(f"Cleaned file saved to '{cleaned_output_file}'")
print(f"Original size: {len(raw_text)} characters")
print(f"Cleaned size: {len(cleaned_text)} characters")
```

```python
# Remove "page..." patterns from the cleaned text
cleaned_text = re.sub(r'page[^\n]*', '', cleaned_text, flags=re.IGNORECASE)

# Remove extra blank lines again after removing page patterns
cleaned_text = re.sub(r'\n\s*\n+', '\n', cleaned_text)
cleaned_text = cleaned_text.strip()

# Write updated content back to the cleaned file
with open(cleaned_output_file, 'w', encoding='utf-8') as outfile:
    outfile.write(cleaned_text)

print(f"Updated '{cleaned_output_file}' - removed 'page...' patterns")
```
